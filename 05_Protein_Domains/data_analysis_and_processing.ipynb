### **Task 3**
We need to determine the correlation coefficient between the expression level of gene A and gene B. Then we need to interpret the obtained result.

```python
gen_a = [0.7, 0.74, 0.90, 1.05, 1.2, 1.31]
gen_b = [1.5, 1.32, 1.1, 1.2, 0.72, 0.60]
```

```python
from scipy import stats

res = stats.spearmanr(gen_a, gen_b)
print(round(res.statistic, 3))
print(round(res.pvalue, 3))
```

```python
import matplotlib.pyplot as plt

plt.scatter(gen_a, gen_b)
plt.xlabel("Gen A")
plt.ylabel("Gen B")
plt.title("Gene Expression Level Plot")
plt.show()
```

From the correlation test, a correlation value of -0.943 was obtained - this means that as the level of gene A increases, the level of gene B decreases. However, this does not mean that we can be sure that an increase in the level of gene B causes a decrease in the level of gene A, or similarly the other way around.

The P-value of this test is 0.005, which, assuming an alpha level of 5%, confirms the significance of the obtained correlation.
