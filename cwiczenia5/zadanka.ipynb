### **Zadanie 2**
W oparciu o metodę SVM zbudowano model w postaci:

```python
def model(a, b, c):
    return 0.6*a - 0.02*b - 0.4*c
```

Według osoba o nastęującym profilu ekspresji genów jest chora:

```python
chory = model(0.8, 0.7, 0.2)
print(round(chory, 2))
```

Oznacza to, ze osoby ktore są powyzej zera są traktowane jako chore. Więc dla nowej osoby, musimy sprawdzić czy wynik dla osoby jest powyzej, czy ponizej zera.

Czy osoba o profilu ekspresji (A, B, C) = (0.8, 0.7, 0.2) jest chora?

```python
osoba = model(0.6, 0.9, 0.4)
print(round(osoba, 2))

if osoba < 0:
    print("Osoba jest zdrowa")
else:
    print("Osoba jest chora")
```

```python
def model_rekurencyjny(a, c):
    return 0.6*a - 0.4*c
```

```python
osoba_rek = model_rekurencyjny(0.6, 0.4)
print(round(osoba_rek, 2))

if osoba < 0:
    print("Osoba jest zdrowa")
else:
    print("Osoba jest chora")
```

### **Zadanie 3**
Musimy wyznaczyć współczynnik korelacji pomiędzy poziomem ekspresji genu A i genu B. Nastepnie trzeba zinterpretować uzyskany wynik.

```python
gen_a = [0.7, 0.74, 0.90, 1.05, 1.2, 1.31]
gen_b = [1.5, 1.32, 1.1, 1.2, 0.72, 0.60]
```

```python
from scipy import stats

res = stats.spearmanr(gen_a, gen_b)
print(round(res.statistic, 3))
print(round(res.pvalue, 3))
```

```python
import matplotlib.pyplot as plt

plt.scatter(gen_a, gen_b)
plt.xlabel("Gen A")
plt.ylabel("Gen B")
plt.title("Wykres poziomu ekspresji genów")
plt.show()
```

Z testu korelacji otrzymano wartość korelacji na poziomime -0.943 - oznacza to, ze wraz ze wzrostem poziomu genu A, maleje poziom genu B. Nie powoduje to natomiast, ze mozemy być pewni, ze wzrost poziomu genu B powoduje spadek poziomu genu A, ani analogicznie w drugą stronę.

P-value tego testu wynosi 0,005, co przy załozeniu alfa na poziomie 5%, potwierdza istotność otrzymanej korelacji.
