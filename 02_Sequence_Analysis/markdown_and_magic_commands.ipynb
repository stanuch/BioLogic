### **7. _%%bash_**

<font size=2>The following is the code in Bash (text shell, <i>shell</i>, Linux):
```Bash
for i in {0..5}
to
    echo "Next number: ${i}"
done
```

It is equivalent to Python's `for` loop:
```Python
for i in range(6):
    print(f'Next number: {i}')
```

Using the `%%bash` command, run the above Bash snippet in the cell below.</font>

````python
%%bash
for i in {0..5}
do
    echo "Next number: ${i}"
done
````

<font size=2>In the next two cells, run the same code again, capturing its output (<i>standard output</i>, <i>stdout</i>) into the `proc` variable with the `%%capture` command and then displaying it on the screen using the Python `print()` function.</font>

````python
%%capture proc
%%bash
for i in {0..5}
do
    echo "Next number: ${i}"
done
````

````python
print(proc.stdout)
````

### **8. _!_ and _!!_**

<font size=2>A single Bash expression embedded in Python code can be run by preceding it with an exclamation point (`!`). When we precede them with two (`!!`), the effect of the code will be saved to the `_` variable. However, it is possible to achieve the same effect by assigning the value of an expression preceded by one exclamation point (`!`) to a variable.
\
\
Bash code fragments are processed by the IPython interpreter before they are sent to the Bash interpreter. Therefore, you can use the values ​​of Python variables by entering them in curly brackets, e.g.: `{variable}`.
\
\
Here you should take into account that the Bash language uses syntactically similar expressions, e.g.: `echo "Working directory: ${PWD}"` (you can try running it in a text terminal). If we don't want unpredictable behavior here, we must double-wrap the braces and include each reference to the Bash variable in the `${{...}}` construct.
\
\
Run and analyze the following code.</font>

````python
# display text with Bash command 'echo'

! echo "Hej!"
````

````python
# display the PWD (print working directory) system variable,
# i.e. the current working directory, with a Bash expression:

! echo ${PWD}

print('\nlub...\n')

! echo ${PWD}
````

````python
# we define the PWD variable in Python, and
# we run the same Bash command,
# getting completely different behavior

PWD = 'Hej'

! echo $PWD
````

````python
# similarly here

! echo ${PWD}
````

````python
# this code clearly indicates what we want to do

! echo ${{PWD}}
````

````python
# similarly here

! echo "Python variable: {PWD}"

! echo "Bash variable: ${{PWD}}"
````

````python
# przypisanie do zmiennej '_'

!! echo ${{PWD}}
````

````python
print(_)
````

````python
# or slightly differently

stdout_lines = ! echo ${{PWD}}
````

````python
print(stdout_lines)
````
