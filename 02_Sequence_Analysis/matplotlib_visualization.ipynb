### **2. Generating data for the chart**
<font size=2>Follow the instructions in the descriptions and comments in the cells to create a simple line chart. Use the information provided in the lecture presentation to help you.
    
Get to know the functions you will work with:
    

| Function type | |
|-|-|
| square | $y(x) = 0.5 \cdot x^{2} - 2.5 \cdot x + 0.5$ |
| sinusoidal | $y(x) = 8.5 \cdot sin(x)$ |
| and their sum | $y(x) = 0.5 \cdot x^{2} - 2.5 \cdot x + 0.5 + 8.5 \cdot sin(x)$ |

</font>

```python
# run the code below,
# so that Jupyter does not change the margins of the chart,
# whose elements are outside the graphics,
# so that the graph in the notebook will look like this
# exactly like the one saved to the file

%config InlineBackend.print_figure_kwargs = {'bbox_inches':None}
```

```python
# import numpy module with name 'np'
# and matplotlib.pyplot under the name 'plt'

import numpy as np
import matplotlib.pyplot as plt
```

```python
# in this cell:
# - create a vector X of values ​​from -10.0 to 15.0 every 0.2
# - create a Y_square vector which is a vector of values ​​of a quadratic function
# - create a vector of Y_sin values, which is a vector of sine function values,
# - use np.sin(...) function
# - create a Y vector that is the sum of Y_square and Y_sin

#X = ...
x = np.arange(-10.0, 15.0, 0.2)

#Y_square = ...
y_square = x**2
#Y_sin = ...
y_sin = np.sin(x)

#Y = ...
y = y_square + y_sin
```

### **3. Simple line chart**

<font size=2>
    
Draw a `Y(X)` chart, completely basic, 5x4 inches in size and 150 dpi. Then save it to the file `output/chart_1.png` with an opaque background and at 300 dpi.
    
</font>

```python
fig = plt.figure(figsize=(5,4), dpi=150)
ax = fig.add_axes([0.15, 0.15, 0.8, 0.8])

ax.plot(x,y)
fig.savefig('output/chart_1.png', transparent=False, dpi=300)
```

### **4. Format a line chart**

<font size=2>A description of the methods specific to the axes object (Axes, `ax`) that you will have to use can be found [here](https://matplotlib.org/3.1.1/api/axes_api.html#axis-limits).
\
\
Copy the code from the previous cell and <u>delete the line of code that draws the line chart</u>.
\
\
Within the created figure and on the same axes, draw three different graphs:
    
1. a red (`color` parameter) line with a thickness (`linewidth`) of 0.8 points and an transparency (`alpha`) of 0.5, draw a `Y_square(X)` graph;
1. with a blue line with a thickness of 0.8 points and an transparency of 0.5, draw a graph of `Y_sin(X)`;
1. Draw the `Y(X)` chart in the form of round points (`marker='o'`) with a size (`markersize`) of 3 points, with a fill color (`markerfacecolor`) of red, a border color (`markeredgecolor`) of blue, a border thickness (`markeredgewidth`) of 0.5 points, without lines (`linestyle=' '`)

Add a legend to the chart as a list of labels with the values `Square, Sin, Sum` using the `ax.legend(...)` method.

Then add the chart title and axis labels (e.g. x and y) to the axis using the `ax.set_title(...), ax.set_xlabel(...), ax.set_ylabel(...)` methods. Set the title to the full mathematical formula of the `Y(X)` function using the expression $\LaTeX$. In addition to the information contained in notebook 0, it will be helpful to know that in $\LaTeX$ expressions, the dot denoting multiplication is `\cdot`.
\
\
If the chart does not fit within the graphic, adjust the position of the axes in the call to the `fig.add_axes(...)` method.
\
\
Save the chart to the file `output/chart_2.png`.
\
\
</font>

```python

fig = plt.figure(figsize=(5,5), dpi=150)
ax = fig.add_axes([0.15, 0.15, 0.8, 0.8])

ax.plot(x,y_square, color='red', linewidth=0.8, alpha=0.5)
ax.plot(x,y_sin, color='blue', linewidth=0.8, alpha=0.5)
ax.plot(x,y, marker='o', markersize=3, markerfacecolor='red', markeredgecolor='blue', markeredgewidth=0.5, linestyle=' ')

ax.legend(['Square', 'Sin', 'Sum'])

ax.set_title(r'$Y(X) = X^2 + \sin(X)$')
ax.set_xlabel('X')
ax.set_ylabel('Y')

plt.savefig('output/chart_2.png')
```
